# EDA: First Touchdown Distributions

```python
import sys
sys.path.insert(0, '..')
import polars as pl
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')

first_tds = pl.read_parquet('../data/processed/first_tds.parquet')
candidates = pl.read_parquet('../data/processed/candidates.parquet')
print(f'First TDs: {first_tds.shape}')
print(f'Candidates: {candidates.shape}')
```

```python
# TD type distribution
td_types = first_tds.group_by('td_type').len().sort('len', descending=True)
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(td_types['td_type'].to_list(), td_types['len'].to_list())
ax.set_title('First TD Type Distribution')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# First TD by position (from candidates where target=1)
scorers = candidates.filter(pl.col('target') == 1)
pos_dist = scorers.group_by('position').len().sort('len', descending=True)
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(pos_dist['position'].to_list(), pos_dist['len'].to_list())
ax.set_title('First TD Scorer by Position')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Trend by season
by_season = first_tds.group_by('season').len().sort('season')
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(by_season['season'].to_list(), by_season['len'].to_list(), marker='o')
ax.set_title('First TDs per Season')
ax.set_xlabel('Season')
ax.set_ylabel('Games')
plt.tight_layout()
plt.show()
```

```python
# Game seconds remaining when first TD occurs
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(first_tds['game_seconds_remaining'].to_list(), bins=50)
ax.set_title('When Does the First TD Happen?')
ax.set_xlabel('Game Seconds Remaining')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```
